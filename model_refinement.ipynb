```python
import pandas as pd
```

```python
df = pd.read_csv('feature_engineer_df.csv', index_col = 'id')
```

```python
test_df = pd.read_csv('feature_engineer_test.csv', index_col='id')
```

```python
df.sample(10)
```

```python
import numpy as np
df['average_tp/category'] = df.groupby(['product_category'])['total_sales'].transform('mean')
df['average_store_tp/category'] = df.groupby(['product_category', 'store_code'])['total_sales'].transform('mean')
group = df.groupby(['product_category', 'store_code'])['total_sales']
df['store_tp/category_SE'] = group.transform('std') / np.sqrt(group.transform('count'))

df['average_tp/product'] = df.groupby(['product_code'])['total_sales'].transform('mean')
```

```python
# Category-level average
average_tp_category = (
    df.groupby('product_category')['total_sales']
      .mean()
)

test_df['average_tp/category'] = (
    test_df['product_category'].map(average_tp_category)
)


average_tp_product = (
    df.groupby('product_code')['total_sales']
      .mean()
)

test_df['average_tp/product'] = (
    test_df['product_code'].map(average_tp_product)
)

# Store × category average
group = df.groupby(
    ['product_category', 'store_code']
)['total_sales']

average_store_tp_category = group.mean()

test_df['average_store_tp/category'] = (
    test_df.set_index(
        ['product_category', 'store_code']
    ).index.map(average_store_tp_category)
)


# Store × category standard error
store_tp_category_SE = (
    group.std() / np.sqrt(group.count())
)

test_df['store_tp/category_SE'] = (
    test_df.set_index(
        ['product_category', 'store_code']
    ).index.map(store_tp_category_SE)
)
```

```python
test_df.sample(10)
```

```python
df.sample(10)
```

```python
X_train = df.drop(columns=['total_sales', 'fat_content']).copy()
y_train = df['total_sales'].copy()

X_test = test_df.drop(columns=['fat_content']).copy()
```

```python
from catboost import CatBoostRegressor
from sklearn.model_selection import KFold, cross_val_score
import numpy as np

categorical_features = [
    'product_code',
    'product_category',
    'store_code',
    'store_size',
    'store_location_tier',
    'store_format'
]

model = CatBoostRegressor(
    iterations=500,
    learning_rate=0.02,
    early_stopping_rounds=100,
    depth=6,
    loss_function='RMSE',
    random_seed=62,
    verbose=0
)

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scores = []

for train_index, validation_index in kf.split(X_train):

    X_fold_train = X_train.iloc[train_index]
    X_fold_valid = X_train.iloc[validation_index]

    y_fold_train = y_train.iloc[train_index]
    y_fold_valid = y_train.iloc[validation_index]

    model.fit(
        X_fold_train,
        y_fold_train,
        cat_features=categorical_features
    )

    predictions = model.predict(X_fold_valid)

    rmse = np.sqrt(np.mean((y_fold_valid - predictions) ** 2))

    scores.append(rmse)

    print(f"Fold RMSE: {rmse:.4f}")

print("\nCross-validation results")
print(f"RMSE scores: {scores}")
print(f"Mean RMSE: {np.mean(scores):.4f}")
print(f"Std RMSE: {np.std(scores):.4f}")
```

```python
final_model = CatBoostRegressor(
    iterations=500,
    learning_rate=0.02,
    early_stopping_rounds=100,
    depth=6,
    loss_function='RMSE',
    random_seed=62,
    verbose=100
)

final_model.fit(
    X_train,
    y_train,
    cat_features=categorical_features
)

y_pred = final_model.predict(X_test)

submission_df = pd.DataFrame({
    'id': test_df.index,
    'total_sales': y_pred
})
submission_df.to_csv('submission_cv_v12_1.csv', index=False)
```
